## 1. 환경 설정

```python
from dotenv import load_dotenv
import requests
import os

load_dotenv()

CLIENT_ID = os.getenv("CLIENT_ID")
CLIENT_SECRET = os.getenv("CLIENT_SECRET")

print("Client ID:", CLIENT_ID[:5] + "*****" if CLIENT_ID else "❌ 없음")
print("Client Secret:", CLIENT_SECRET[:5] + "*****" if CLIENT_SECRET else "❌ 없음")
```

```python
import subprocess
subprocess.run(["pip", "install", "python-dotenv", "requests"])
```

```python
# 키 값 직접 확인
print("CLIENT_ID:", repr(CLIENT_ID))
print("CLIENT_SECRET:", repr(CLIENT_SECRET))
```

## 2. API 검색 테스트

```python
def search_naver(query, display=5):
    url = "https://openapi.naver.com/v1/search/shop.json"
    headers = {
        "X-Naver-Client-Id": CLIENT_ID,
        "X-Naver-Client-Secret": CLIENT_SECRET,
    }
    params = {"query": query, "display": display, "start": 1}
    res = requests.get(url, headers=headers, params=params, timeout=10)
    res.raise_for_status()
    return res.json()

data = search_naver("원통형 텀블러", display=5)
items = data.get("items", [])
print(f"검색 결과: {len(items)}개")
for i, item in enumerate(items):
    print(f"[{i}] {item['title']}")
```

## 3. 이미지 URL 확인

```python
for i, item in enumerate(items):
    print(f"[{i}] {item['image']}")
```

## 4. 이미지 1장 다운로드 테스트

```python
os.makedirs("data/train/straight", exist_ok=True)

img_url = items[0]['image']
save_path = "data/train/straight/test_00.jpg"

res = requests.get(img_url, timeout=5)
with open(save_path, "wb") as f:
    f.write(res.content)

print(f"저장 완료: {save_path}")
print(f"파일 크기: {os.path.getsize(save_path)} bytes")
```

## 5. 저장된 이미지 확인

```python
from IPython.display import Image, display

display(Image(filename=save_path, width=300))
```

## 6. 5장 한꺼번에 저장

```python
for i, item in enumerate(items):
    img_url = item['image']
    save_path = f"data/train/straight/test_{i:02d}.jpg"
    try:
        res = requests.get(img_url, timeout=5)
        with open(save_path, "wb") as f:
            f.write(res.content)
        print(f"[{i}] ✅ 저장: {save_path}")
    except Exception as e:
        print(f"[{i}] ❌ 실패: {e}")
```

## 7. 저장된 이미지 전체 확인

```python
from IPython.display import Image, display
import glob

files = glob.glob("data/train/straight/test_*.jpg")
print(f"저장된 파일: {len(files)}장")
for f in files:
    print(f)
    display(Image(filename=f, width=200))
```
